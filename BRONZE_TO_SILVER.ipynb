```python
dbutils.widgets.text("table_name", "")
table_name = dbutils.widgets.get("table_name")
```

```python
bronze_df = spark.read.table(f"hospital_data_landing.bronze.{table_name}")
```

```python
import pyspark.sql.functions as f
updated_df = bronze_df.withColumn("insert_timestamp",f.current_timestamp())\
.withColumn("updated_timestamp",f.current_timestamp())
```

```python
updated_df.write.mode("overwrite").saveAsTable(f"hospital_data_landing.silver.{table_name}")
```
