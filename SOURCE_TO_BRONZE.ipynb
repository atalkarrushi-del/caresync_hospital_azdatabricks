```python
df=spark.read.format("parquet").load("/Volumes/hospital_data_landing/landing_schema/vol/data_f3d33faf-c8ac-48f5-a110-ae8ba3fd54ad_b0cc74a7-e63d-4ed5-b6cc-662760adadea.parquet")
```

```python
display(df)
```

```python
from pyspark.sql.functions import current_timestamp

landing_df = df.withColumn("landing_timestamp", current_timestamp()) \
       .withColumn("inserted_timestamp", current_timestamp())
```

```python
display(landing_df)
```

```python
landing_df.write.mode("append").saveAsTable("hospital_data_landing.bronze.hospitals")
```
